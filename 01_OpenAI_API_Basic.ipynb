## 1. OpenAI API란?

API(Application Programming Interface)는 한 프로그램이 다른 프로그램의 기능이나 데이터를 사용할 수 있게 연결하는 방식입니다.

### ChatGPT와 OpenAI API의 차이

| 구분 | ChatGPT | OpenAI API |
|---|---|---|
| 사용 방법 | 웹이나 앱에서 직접 대화 | Python 등의 프로그램에서 호출 |
| 입력 | 채팅창에 입력 | 코드의 `input`으로 전달 |
| 출력 | ChatGPT 화면 | Python 변수로 반환 |
| 활용 | 개인 질문과 작업 | 챗봇·분석 도구·업무 자동화 |
| 비용 | ChatGPT 요금제 | API 사용량에 따라 별도 과금 |

> ChatGPT 유료 구독과 OpenAI API 결제는 별도로 운영됩니다.

## 2. 주요 구성요소

| 구성요소 | 의미 |
|---|---|
| API Key | API 사용 권한을 확인하는 비밀키 |
| SDK | Python에서 API를 편리하게 사용하는 라이브러리 |
| Client | Python과 OpenAI를 연결하는 객체 |
| Model | 입력을 처리하고 답변을 생성하는 AI 모델 |
| Instructions | AI가 따라야 하는 역할과 응답 원칙 |
| Input | AI에게 전달하는 질문이나 데이터 |
| Response | OpenAI가 반환한 전체 응답 객체 |
| Output text | 응답에서 추출한 최종 답변 텍스트 |
| Token | 입력과 출력의 사용량을 계산하는 단위 |

질문과 답변이 길어질수록 토큰 사용량과 비용이 늘어날 수 있습니다.

## 7. Responses API 기본 구조

이번 실습은 OpenAI의 통합형 인터페이스인 Responses API를 사용합니다.

```python
from openai import OpenAI
client = OpenAI(api_key=api_key)
MODEL = "gpt-4o-mini"

response = client.responses.create(
    model=MODEL,
    input="AI에게 보낼 질문"
)

print(response.output_text)
```

| 항목 | 의미 |
|---|---|
| `client.responses.create()` | OpenAI에 요청 전송 |
| `model` | 사용할 AI 모델 |
| `input` | AI에게 전달할 질문이나 데이터 |
| `response` | OpenAI가 반환한 전체 응답 |
| `response.output_text` | AI가 생성한 최종 텍스트 |

### 주요 입력값

| 입력값 | 의미 |
|---|---|
| `model` | 사용할 AI 모델 |
| `input` | 질문이나 데이터 |
| `instructions` | 역할과 응답 원칙 |
| `max_output_tokens` | 최대 출력 토큰 수 |
| `stream` | 응답을 실시간으로 받을지 여부 |
| `previous_response_id` | 이전 응답과 연결하여 문맥 유지 |

기본 실습에서는 `model`, `input`, `instructions`를 중심으로 사용합니다.

## 8. 모델 선택

모델을 선택할 때는 답변 품질, 처리 속도, 비용, 작업의 복잡도를 고려합니다.

| 모델 | 특징 | 권장 용도 |
|---|---|---|
| `gpt-5.6-sol` | 높은 성능의 플래그십 모델 | 복잡한 분석·전문 업무·코딩 |
| `gpt-5.6-terra` | 성능과 비용의 균형 | 일반적인 업무 서비스 |
| `gpt-5.6-luna` | 비용 효율적인 모델 | 기본 실습·간단한 챗봇·대량 처리 |

이번 과정에서는 `gpt-4o-mini`를 사용합니다. 모델과 가격은 변경될 수 있으므로 실제 프로젝트에서는 공식 모델 문서를 확인합니다.

## 9. 생성형 AI 사용 시 주의사항

- **환각:** 사실이 아닌 내용을 사실처럼 생성할 수 있음
- **최신성:** 최신 정보가 포함되지 않을 수 있음
- **편향:** 학습 데이터의 편향이 답변에 반영될 수 있음
- **개인정보:** 개인정보와 회사 기밀을 입력하지 않아야 함
- **검증:** 중요한 내용은 반드시 사람이 다시 확인해야 함

> AI의 답변은 완성된 정답이 아니라 사람이 검토하고 활용해야 하는 결과입니다.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv() 
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI()

goal = "데이터 엔지니어를 위한 코딩"
daily_time = 30
current_level = "초급"
difficulty = "중급"

prompt = f'''
목표 : {goal}
하루 가능 시간{daily_time}분
현재 수준 : {current_level}
원하는 수준 : {difficulty}
'''
#print(prompt)

MODEL = 'gpt-4o-mini'

response = client.responses.create(
    model=MODEL,
    instructions='너는 내 코딩공부 담당 파트너야. 나는 월요일부터 일요일까지의 계획이 필요하고 꾸준히 하는걸 버거워해 일일계획이랑 소요시간, 중간점검할 계획을 표형태로 생성' , 
    input=prompt
)

print(response.output_text)
````
